```python
!qiime tools import \
    --type 'SampleData[PairedEndSequencesWithQuality]' \
    --input-path 0_data \
    --output-path 0_data/demux.qza \
    --input-format CasavaOneEightSingleLanePerSampleDirFmt

!qiime demux summarize \
    --i-data 0_data/demux.qza \
    --o-visualization 0_data/demux.qzv

!qiime tools view 0_data/demux.qzv
```

```python
!mkdir -p 1_data_trimmed

!qiime cutadapt trim-paired \
    --i-demultiplexed-sequences 0_data/demux.qza \
    --p-cores 8 \
    --p-front-f CAAGRGTTHGATYMTGGCTCAG \
    --p-front-r TGCTGCCTCCCGTAGGAGT \
    --o-trimmed-sequences 1_data_trimmed/cutadapt.qza \
    --p-match-adapter-wildcards \
    --p-discard-untrimmed \
    --p-match-read-wildcards \
    --verbose
    
# Visualization
!qiime demux summarize \
    --i-data 1_data_trimmed/cutadapt.qza \
    --o-visualization 1_data_trimmed/cutadapt.qzv
```

```python
!mkdir -p 2_denoised

!qiime dada2 denoise-paired \
  --i-demultiplexed-seqs 1_data_trimmed/cutadapt.qza \
  --p-trunc-len-f 200 \
  --p-trunc-len-r 200 \
  --p-n-threads 8 \
  --o-table 2_denoised/table.qza \
  --o-representative-sequences 2_denoised/rep-seqs.qza \
  --o-denoising-stats 2_denoised/dada2-stats.qza

!qiime feature-table summarize \
    --i-table 2_denoised/table.qza \
    --o-visualization 2_denoised/table.qzv

!qiime metadata tabulate \
    --m-input-file 2_denoised/dada2-stats.qza \
    --o-visualization 2_denoised/dada2-stats.qzv

!qiime feature-table tabulate-seqs \
    --i-data 2_denoised/rep-seqs.qza \
    --o-visualization 2_denoised/rep-seqs.qzv
```

```python
# testing effect of other parameters on the percentage of passed filtering
!mkdir -p 2_denoised_test

!qiime dada2 denoise-paired \
  --i-demultiplexed-seqs 1_data_trimmed/cutadapt.qza \
  --p-trunc-len-f 220 \
  --p-trunc-len-r 230 \
  --p-n-threads 8 \
  --o-table 2_denoised_test/table.qza \
  --o-representative-sequences 2_denoised_test/rep-seqs.qza \
  --o-denoising-stats 2_denoised_test/dada2-stats.qza

!qiime feature-table summarize \
    --i-table 2_denoised_test/table.qza \
    --o-visualization 2_denoised_test/table.qzv

!qiime metadata tabulate \
    --m-input-file 2_denoised_test/dada2-stats.qza \
    --o-visualization 2_denoised_test/dada2-stats.qzv

!qiime feature-table tabulate-seqs \
    --i-data 2_denoised_test/rep-seqs.qza \
    --o-visualization 2_denoised_test/rep-seqs.qzv
```

```python
!mkdir -p 3_denoised_filtered

!qiime feature-table filter-features \
    --i-table 2_denoised/table.qza \
    --p-min-frequency 10 \
    --p-min-samples 2 \
    --o-filtered-table 3_denoised_filtered/filt-table.qza

!qiime feature-table filter-seqs \
    --i-data 2_denoised/rep-seqs.qza \
    --i-table 3_denoised_filtered/filt-table.qza \
    --o-filtered-data 3_denoised_filtered/rep-seqs.qza

# Visualization
!qiime feature-table summarize \
    --i-table 3_denoised_filtered/filt-table.qza \
    --o-visualization 3_denoised_filtered/filt-table.qzv
```

```python
!mkdir -p 4_taxonomy

!qiime feature-classifier classify-hybrid-vsearch-sklearn \
  --i-query 3_denoised_filtered/rep-seqs.qza \
  --i-reference-reads classifier/seqs-V12.qza \
  --i-reference-taxonomy classifier/taxa-V12.qza \
  --i-classifier classifier/classifier-V12.qza \
  --o-classification 4_taxonomy/taxonomy.qza \
  --p-threads 8 \
  --p-no-prefilter

!qiime metadata tabulate \
    --m-input-file 4_taxonomy/taxonomy.qza \
    --o-visualization 4_taxonomy/taxonomy.qzv
```

```python
#!mkdir -p 5_tree

!qiime phylogeny align-to-tree-mafft-fasttree \
    --i-sequences 3_denoised_filtered/rep-seqs.qza \
    --p-n-threads 8 \
    --output-dir 5_tree
```

```python
!mkdir -p 6_diversity

!qiime diversity alpha-rarefaction \
    --i-table 3_denoised_filtered/filt-table.qza \
    --i-phylogeny 5_tree/rooted_tree.qza \
    --p-max-depth 10300 \
    --m-metadata-file meta.txt \
    --o-visualization 6_diversity/rarefaction.qzv

!qiime diversity core-metrics-phylogenetic \
    --i-phylogeny 5_tree/rooted_tree.qza  \
    --i-table 3_denoised_filtered/filt-table.qza \
    --p-sampling-depth 10300 \
    --p-n-jobs-or-threads 8 \
    --m-metadata-file meta.txt \
    --output-dir 6_diversity/Core-metrics
```
